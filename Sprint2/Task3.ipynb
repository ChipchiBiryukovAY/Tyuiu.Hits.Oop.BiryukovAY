[Описание задачи:
Создать базовый класс Student в C#, который будет представлять информацию о студентах. На основе этого класса разработать 2-3 производных класса, демонстрирующих принципы наследования и полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и методы, а также переопределены некоторые методы базового класса для демонстрации полиморфизма.
Требования к базовому классу Student:
•
Атрибуты: Имя (Name), Возраст (Age), Курс (Course).
•
Методы:
o
GetInfo(): метод для получения информации о студенте в виде строки.
o
Study(): метод для вывода сообщения о том, что студент учится.
o
TakeExam(): метод для вывода сообщения о сдаче экзамена.
Требования к производным классам:
1.
Студент бакалавриата (BachelorStudent): Должен содержать дополнительные атрибуты, такие как Специальность (Specialty). Метод Study() должен быть переопределен для добавления информации о специальности при изучении предмета.
2.
Студент магистратуры (MasterStudent): Должен содержать дополнительные атрибуты, такие как Научный руководитель (ScientificAdvisor). Метод TakeExam() должен быть переопределен для добавления информации о научном руководителе при сдаче экзамена.
3.
Аспирант (PhDStudent) (если требуется третий класс): Должен содержать дополнительные атрибуты, такие как Тема диссертации (DissertationTopic). Метод Study() должен быть переопределен для добавления информации о теме диссертации при изучении предмета.

Добавьте к сущестующим классам конструктора классов с использованием гетторов и сетторов и реализуйте взаимодействие объектов между собой]

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте простое, сложное и множественное наследование

<h2 style="color:DodgerBlue">Реализация:</h2>

```polyglot-notebook
public interface IResearchable
{
    void Research();
}

public interface IInternship
{
    void DoInternship();
}

public interface ITeachable
{
    void Teach();
}

public class Student
{
    private string _name;
    private int _age;
    private int _course;
    private string _id;          
    private string _email;       
    private double _gpa;         
    private int _enrollmentYear; 

    public string Name { get { return _name; } set { _name = value; } }
    public int Age
    {
        get { return _age; }
        set { if (value >= 0) _age = value; else Console.WriteLine("Возраст не может быть отрицательным!"); }
    }
    public int Course
    {
        get { return _course; }
        set { if (value >= 1) _course = value; else Console.WriteLine("Курс должен быть >= 1"); }
    }
    public string Id { get { return _id; } set { _id = value; } }
    public string Email { get { return _email; } set { _email = value; } }
    public double Gpa
    {
        get { return _gpa; }
        set { if (value >= 0 && value <= 5) _gpa = value; else Console.WriteLine("GPA от 0 до 5"); }
    }
    public int EnrollmentYear { get { return _enrollmentYear; } set { _enrollmentYear = value; } }

    public static int TotalStudents = 0;

    public Student(string name, int age, int course, string id, string email, double gpa, int year)
    {
        Name = name;
        Age = age;
        Course = course;
        Id = id;
        Email = email;
        Gpa = gpa;
        EnrollmentYear = year;
        TotalStudents++;
    }

    public virtual string GetInfo()
    {
        return $"Студент: {Name}, возраст: {Age}, курс: {Course}, ID: {Id}, GPA: {Gpa}";
    }

    public virtual void Study()
    {
        Console.WriteLine($"{Name} учится");
    }

    public virtual void TakeExam()
    {
        Console.WriteLine($"{Name} сдаёт экзамен");
    }

    public void UpdateGpa(double newGpa)
    {
        Gpa = newGpa;
        Console.WriteLine($"{Name} обновил GPA: {Gpa}");
    }

    public void SendEmail(string message)
    {
        Console.WriteLine($"Письмо для {Email}: {message}");
    }

    public bool IsExcellentStudent()
    {
        return Gpa >= 4.5;
    }

    public static void ShowTotal()
    {
        Console.WriteLine($"Всего студентов: {TotalStudents}");
    }
}

public class GraduateStudent : Student
{
    public string ScientificAdvisor { get; set; }
    public string ThesisTopic { get; set; }

    public GraduateStudent(string name, int age, int course, string id, string email, double gpa, int year,
                           string advisor, string topic)
        : base(name, age, course, id, email, gpa, year)
    {
        ScientificAdvisor = advisor;
        ThesisTopic = topic;
    }

    public override void Study()
    {
        Console.WriteLine($"{Name} изучает тему: {ThesisTopic} (науч. рук.: {ScientificAdvisor})");
    }

    public void ConsultWithAdvisor()
    {
        Console.WriteLine($"{Name} консультируется с {ScientificAdvisor}");
    }
}

public class BachelorStudent : Student, IInternship
{
    public string Specialty { get; set; }
    public string University { get; set; }
    public string InternshipCompany { get; set; }

    public BachelorStudent(string name, int age, int course, string id, string email, double gpa, int year,
                           string specialty, string university)
        : base(name, age, course, id, email, gpa, year)
    {
        Specialty = specialty;
        University = university;
    }

    public override void Study()
    {
        Console.WriteLine($"{Name} изучает {Specialty} в {University}");
    }

    public void DoInternship()
    {
        Console.WriteLine($"{Name} проходит стажировку в {InternshipCompany}");
    }
}

public class MasterStudent : GraduateStudent, IResearchable, IInternship
{
    public int ConferencesCount { get; set; }

    public MasterStudent(string name, int age, int course, string id, string email, double gpa, int year,
                         string advisor, string topic, int conferences)
        : base(name, age, course, id, email, gpa, year, advisor, topic)
    {
        ConferencesCount = conferences;
    }

    public override void TakeExam()
    {
        Console.WriteLine($"{Name} сдаёт экзамен под руководством {ScientificAdvisor}");
    }

    public void Research()
    {
        Console.WriteLine($"{Name} исследует тему {ThesisTopic}");
    }

    public void DoInternship()
    {
        Console.WriteLine($"{Name} на научной стажировке");
    }
}

public class PhDStudent : GraduateStudent, IResearchable, ITeachable
{
    public int PublicationsCount { get; set; }
    public int TeachingHours { get; set; }

    public PhDStudent(string name, int age, int course, string id, string email, double gpa, int year,
                      string advisor, string topic, int publications, int hours)
        : base(name, age, course, id, email, gpa, year, advisor, topic)
    {
        PublicationsCount = publications;
        TeachingHours = hours;
    }

    public override void Study()
    {
        Console.WriteLine($"{Name} изучает диссертацию: {ThesisTopic}");
    }

    public override void TakeExam()
    {
        Console.WriteLine($"{Name} сдаёт кандидатский экзамен");
    }

    public void Research()
    {
        Console.WriteLine($"{Name} публикует статьи по теме {ThesisTopic}. Публикаций: {PublicationsCount}");
    }

    public void Teach()
    {
        Console.WriteLine($"{Name} ведёт семинары ({TeachingHours} часов)");
    }

    public void PublishArticle(string title)
    {
        PublicationsCount++;
        Console.WriteLine($"{Name} опубликовал статью: {title}. Всего: {PublicationsCount}");
    }
}

BachelorStudent b = new BachelorStudent("Mike", 18, 1, "B001", "mike@uni.ru", 4.5, 2025, "Программирование", "ТИУ");
MasterStudent m = new MasterStudent("Monica", 23, 1, "M001", "monica@uni.ru", 4.8, 2024, "Иванов И.И.", "ИИ", 3);
PhDStudent p = new PhDStudent("Евгений", 26, 1, "P001", "evgeny@uni.ru", 4.9, 2023, "Петров П.П.", "ML", 7, 120);

b.InternshipCompany = "Яндекс";
b.University = "ТИУ";

Console.WriteLine(b.GetInfo());
b.Study();
b.TakeExam();
b.DoInternship();
Console.WriteLine("---");

Console.WriteLine(m.GetInfo());
m.Study();
m.TakeExam();
m.Research();
m.DoInternship();
m.ConsultWithAdvisor();
Console.WriteLine("---");

Console.WriteLine(p.GetInfo());
p.Study();
p.TakeExam();
p.Research();
p.Teach();
p.PublishArticle("Новый метод обучения");
p.ConsultWithAdvisor();
Console.WriteLine("---");

b.UpdateGpa(5.0);
b.SendEmail("Поздравляем с отличной сессией!");
Console.WriteLine($"{b.Name} отличник? {b.IsExcellentStudent()}");

Student[] students = { b, m, p };
Console.WriteLine("\n=== Полиморфизм ===");
foreach (Student s in students)
{
    Console.WriteLine(s.GetInfo());
    s.Study();
    s.TakeExam();
    Console.WriteLine("---");
}

Student.ShowTotal();
```
