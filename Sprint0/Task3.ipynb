[Описание задачи:
Создать базовый класс Student в C#, который будет представлять информацию о студентах. На основе этого класса разработать 2-3 производных класса, демонстрирующих принципы наследования и полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и методы, а также переопределены некоторые методы базового класса для демонстрации полиморфизма.
Требования к базовому классу Student:
•
Атрибуты: Имя (Name), Возраст (Age), Курс (Course).
•
Методы:
o
GetInfo(): метод для получения информации о студенте в виде строки.
o
Study(): метод для вывода сообщения о том, что студент учится.
o
TakeExam(): метод для вывода сообщения о сдаче экзамена.
Требования к производным классам:
1.
Студент бакалавриата (BachelorStudent): Должен содержать дополнительные атрибуты, такие как Специальность (Specialty). Метод Study() должен быть переопределен для добавления информации о специальности при изучении предмета.
2.
Студент магистратуры (MasterStudent): Должен содержать дополнительные атрибуты, такие как Научный руководитель (ScientificAdvisor). Метод TakeExam() должен быть переопределен для добавления информации о научном руководителе при сдаче экзамена.
3.
Аспирант (PhDStudent) (если требуется третий класс): Должен содержать дополнительные атрибуты, такие как Тема диссертации (DissertationTopic). Метод Study() должен быть переопределен для добавления информации о теме диссертации при изучении предмета.]


<h2 style="color:DodgerBlue">Реализация:</h2>

----

```polyglot-notebook
public class Student
{
    public string Name { get; set; }     
    public int Age { get; set; }         
    public int Course { get; set; }      

    public Student(string name, int age, int course)
    {
        Name = name;
        Age = age;
        Course = course;
    }

    public virtual string GetInfo()
    {
        return $"Студент: {Name}, возраст: {Age}, курс: {Course}";
    }

    public virtual void Study()
    {
        Console.WriteLine($"{Name} учится");
    }

    public virtual void TakeExam()
    {
        Console.WriteLine($"{Name} сдаёт экзамен");
    }
}
Student s = new Student("Alex", 20, 2);
Console.WriteLine(s.GetInfo());
s.Study();
s.TakeExam();
```

```polyglot-notebook
public class BachelorStudent : Student
{
    public string Specialty { get; set; }

    public BachelorStudent(string name, int age, int course, string specialty)
        : base(name, age, course)   // ← передаём общие данные в родителя
    {
        Specialty = specialty;
    }

    public override void Study()
    {
        Console.WriteLine($"{Name} изучает предметы по специальности {Specialty}");
    }
}
BachelorStudent b = new BachelorStudent("Mike", 18, 1, "Программирование");
b.Study();
b.TakeExam();      
```

```polyglot-notebook
public class MasterStudent : Student
{
    public string ScientificAdvisor { get; set; }

    public MasterStudent(string name, int age, int course, string advisor)
        : base(name, age, course)
    {
        ScientificAdvisor = advisor;
    }

    public override void TakeExam()
    {
        Console.WriteLine($"{Name} сдаёт экзамен под руководством {ScientificAdvisor}");
    }
}
MasterStudent m = new MasterStudent("Monica", 23, 1, "Иванов И.И.");

m.Study();      
m.TakeExam();    
```

```polyglot-notebook
public class PhDStudent : Student
{
    public string DissertationTopic { get; set; }

    public PhDStudent(string name, int age, int course, string topic)
        : base(name, age, course)
    {
        DissertationTopic = topic;
    }

    public override void Study()
    {
        Console.WriteLine($"{Name} изучает материалы по теме диссертации: {DissertationTopic}");
    }
}
PhDStudent p = new PhDStudent("Евгений", 26, 1, "Искусственный интеллект");
p.Study();
p.TakeExam();
```

```polyglot-notebook
Student[] students = {
    new BachelorStudent("Mike", 18, 1, "Программирование"),
    new MasterStudent("Monica", 23, 1, "Иванов И.И."),
    new PhDStudent("Евгений", 26, 1, "Искусственный интеллект")
};

foreach (Student s in students)
{
    Console.WriteLine(s.GetInfo());
    s.Study();
    s.TakeExam();
    Console.WriteLine("---");
}
```
